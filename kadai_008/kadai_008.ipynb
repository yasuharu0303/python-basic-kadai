```python
# ランダムな整数を利用するために、randomモジュールをインポート
import random

# 変数varに1,20までのランダムな整数を代入する
var = random.randint(1,20)

# 変数varが、3と5の両方の倍数の時は「FizzBuzz」,3の倍数の場合は「Fizz」,5の倍数の時は「Buzz」,それ以外は「var」と出力
if       var % 3 == 0 and var % 5 == 0:
         print("FizzBuzz")

elif     var % 3 == 0:
         print("Fizz")
elif     var % 5 == 0:
         print("Buzz")
else:
         print(var)
```
